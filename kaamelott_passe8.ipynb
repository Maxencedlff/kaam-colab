# Kaamelott, passe 8 : empreintes vocales
GPU T4, puis Tout exécuter. À la fin (environ 15 min), `p8_voix.zip` arrive dans Téléchargements.

```
import os, sys, subprocess
D = '/content/kaam8'; os.makedirs(D, exist_ok=True)
subprocess.run(['wget', '-q', '-O', D + '/kaam8.py', 'https://huggingface.co/datasets/Maxencedlf/kaamelott-videos/resolve/main/p8/kaam8.py'], check=True)
sys.path.insert(0, D); import kaam8
kaam8.telecharger(D); kaam8.main(D); kaam8.rendre(D)
```
