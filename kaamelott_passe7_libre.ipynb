# Kaamelott, passe 7 : 2e onglet (transcription « libre »)
GPU T4, puis Tout exécuter. À la fin, `p7_libre.zip` arrive dans Téléchargements.

```
import os, sys, subprocess
D = '/content/kaam7'; os.makedirs(D, exist_ok=True)
subprocess.run(['wget', '-q', '-O', D + '/kaam7.py', 'https://huggingface.co/datasets/Maxencedlf/kaamelott-videos/resolve/main/p7/kaam7.py'], check=True)
sys.path.insert(0, D); import kaam7
kaam7.telecharger(D); kaam7.main(D, passes=('libre',)); kaam7.rendre(D, passes=('libre',))
```
